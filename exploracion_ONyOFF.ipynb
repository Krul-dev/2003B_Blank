```python
import numpy as np
import pandas as pd

def ang_diff_deg(a, b):
    """Diferencia angular mínima entre ángulos a y b en grados."""
    return np.abs((a - b + 180) % 360 - 180)

def separar_on_off_por_estacion(df, est, angulo, tol, ws_col=None, ws_min=None):
    """Separa registros on-wind y off-wind de una estación según un ángulo y tolerancia."""
    # Usar el nombre correcto de la columna de viento para NORTE 2
    if est == "NORTE 2":
        wind_col = "NORTE 2 WDR DEG"
    else:
        wind_col = f"{est} WDV DEG"
    df = df.dropna(subset=[wind_col]).copy()
    df[wind_col] = pd.to_numeric(df[wind_col], errors="coerce")
    df["ang_diff"] = ang_diff_deg(df[wind_col], angulo)
    mask_on = df["ang_diff"] <= tol
    if ws_col and ws_col in df.columns and ws_min is not None:
        df[ws_col] = pd.to_numeric(df[ws_col], errors="coerce")
        mask_on &= df[ws_col] >= ws_min
    on_wind = df[mask_on].drop(columns=["ang_diff"])
    off_wind = df[~mask_on].drop(columns=["ang_diff"])
    return on_wind, off_wind

# Definimos reglas para cada estación
reglas = {
    "NORTE 2":   (315, 45),
    "NORESTE": ( 45, 45),
    "CENTRO":  (225, 45),
}

# Diccionarios para guardar resultados
on_wind_dict = {}
off_wind_dict = {}

for est, (angulo, tol) in reglas.items():
    df_est = dfs_estaciones_clean[est]
    on, off = separar_on_off_por_estacion(df_est, est, angulo, tol)
    on["Estacion"] = est
    off["Estacion"] = est
    on_wind_dict[est] = on
    off_wind_dict[est] = off

# Juntar si quieres un solo DataFrame
on_wind = pd.concat(on_wind_dict.values(), ignore_index=True)
off_wind = pd.concat(off_wind_dict.values(), ignore_index=True)

print("ON-WIND por estación:")
for est, df in on_wind_dict.items():
    print(f"{est}: {len(df)} registros")

print("OFF-WIND por estación:")
for est, df in off_wind_dict.items():
    print(f"{est}: {len(df)} registros")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu

def mannwhitney_daily_median_test(on_wind_dict, off_wind_dict):
    """
    Para cada estación y cada variable (con prefijo de estación), 
    calcula la mediana diaria en ON/OFF wind y aplica Mann–Whitney U test.
    Retorna un DataFrame resumen.
    """
    resultados = []
    for est in on_wind_dict.keys():
        df_on = on_wind_dict[est].copy()
        df_off = off_wind_dict[est].copy()

        # Asegurar que Date sea datetime.date
        df_on['Date'] = pd.to_datetime(df_on['Date']).dt.date
        df_off['Date'] = pd.to_datetime(df_off['Date']).dt.date

        # Variables de esa estación (ignorar Date y Estacion)
        variables_est = [col for col in df_on.columns if col not in ['Date', 'Estacion']]

        for var in variables_est:
            if var not in df_off.columns:
                continue  # por si falta en OFF

            # Mediana diaria
            mediana_on = df_on.groupby('Date')[var].median().dropna()
            mediana_off = df_off.groupby('Date')[var].median().dropna()

            # Mann–Whitney
            if len(mediana_on) > 0 and len(mediana_off) > 0:
                stat, pval = mannwhitneyu(mediana_on, mediana_off, alternative='two-sided')
                diff = mediana_on.median() - mediana_off.median()
            else:
                stat, pval, diff = np.nan, np.nan, np.nan

            # Quitar prefijo de estación para el nombre de variable
            var_clean = var.replace(f"{est} ", "")

            resultados.append({
                'Estacion': est,
                'Variable': var_clean,
                'n_dias_on': len(mediana_on),
                'n_dias_off': len(mediana_off),
                'mediana_on': mediana_on.median() if len(mediana_on) > 0 else np.nan,
                'mediana_off': mediana_off.median() if len(mediana_off) > 0 else np.nan,
                'diff_medianas': diff,
                'U_stat': stat,
                'p_valor': pval
            })
    return pd.DataFrame(resultados)
```

```python

def plot_on_off_means_with_stats(on_wind_dict, off_wind_dict, estacion):
    """
    Genera gráfico de barras con medianas ON/OFF para TODAS las variables de la estación.
    Anota p-value y U-stat encima de cada grupo de barras.
    """
    df = resultado
    df= df[df['Estacion'] == estacion]

    variables = df["Variable"]
    medias_on = df["mediana_on"]
    medias_off = df["mediana_off"]
    p_values = df["p_valor"]
    u_stats = df["U_stat"]

    x = range(len(variables))
    width = 0.35

    fig, ax = plt.subplots(figsize=(14, 6))
    bars_on = ax.bar([i - width/2 for i in x], medias_on, width, label="ON WIND", color="skyblue")
    bars_off = ax.bar([i + width/2 for i in x], medias_off, width, label="OFF WIND", color="salmon")

    # Anotar p-value y U-stat
    for i, (bar_on, bar_off, p, u) in enumerate(zip(bars_on, bars_off, p_values, u_stats)):
        y_max = max(bar_on.get_height(), bar_off.get_height())
        ax.text(i, y_max + 0.05*y_max, f"p={p:.3g}\nU={u:.0f}", 
                ha="center", va="bottom", fontsize=8, color="black")

    ax.set_xticks(x)
    ax.set_xticklabels(variables, rotation=45, ha="right")
    ax.set_ylabel("Mediana diaria")
    ax.set_title(f"Mediana diaria ON vs OFF WIND - {estacion}")
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
from statsmodels.stats.stattools import durbin_watson
from statsmodels.stats.diagnostic import acorr_ljungbox

def autocorrelation_tests_subsample(on_wind_dict, off_wind_dict, freq=7):
    """
    Aplica submuestreo (cada `freq` días) y evalúa autocorrelación
    con Durbin-Watson y Ljung-Box en las primeras 7 variables de cada estación.
    Retorna un DataFrame resumen.
    """
    resultados = []

    for tipo, dicc in zip(["ON", "OFF"], [on_wind_dict, off_wind_dict]):
        for est, df in dicc.items():
            # Ordenar por fecha y eliminar nulos
            df = df.sort_values("Date").dropna()
            df["Date"] = pd.to_datetime(df["Date"])
            
            # Submuestreo semanal
            df_sub = df.set_index("Date").iloc[::freq]

            # Tomamos solo las primeras 7 variables (excluyendo Date y Estacion)
            variables = df_sub.columns[1:8]

            for var in variables:
                serie = df_sub[var].dropna()

                if len(serie) > 10:  # mínimo de datos
                    # Durbin–Watson
                    dw = durbin_watson(serie)

                    # Ljung–Box
                    lb_test = acorr_ljungbox(serie, lags=[10], return_df=True)
                    pval = lb_test['lb_pvalue'].values[0]

                    resultados.append({
                        "Estacion": est,
                        "Variable": var,
                        "Tipo": tipo,
                        "Durbin_Watson": dw,
                        "LjungBox_pval": pval,
                        "n_dias": len(serie)
                    })

    return pd.DataFrame(resultados)

# --- EJEMPLO DE USO ---
# resultados_sub = autocorrelation_tests_subsample(on_wind_dict, off_wind_dict, freq=7)
# print(resultados_sub.head())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import mannwhitneyu

sns.set(style="whitegrid", font_scale=1.1)

def summary_and_plots_with_stats(scores_dict, factor_cols=["Factor1", "Factor3"], group_col="Condicion",
                                 group_labels=("ON","OFF"), figsize=(13,6), show_bar_ci=True,
                                 save_fig=True, output_prefix="resultados"):
    """
    - df: DataFrame con columnas de factors y columna de grupo ("Condicion") con valores 'ON'/'OFF'
    - factor_cols: lista con los nombres de las columnas de factores a analizar (ej. ["Factor1","Factor3"])
    - group_col: nombre de la columna que indica ON/OFF
    - group_labels: etiquetas esperadas en group_col (ON, OFF)
    - show_bar_ci: si True dibuja un gráfico de barras con medias + IC bootstrap
    - save_fig: si True guarda las figuras como PNG
    - output_prefix: prefijo del nombre de archivo para guardar figuras
    Retorna: resumen DataFrame y muestra/guarda plots
    """
    resultados = []
    # tabla de medias/medianas por grupo
    tabla_medias = df.groupby(group_col)[factor_cols].agg(['mean','median','count'])
    tabla_medias.columns = ['_'.join(col).strip() for col in tabla_medias.columns.values]

    # --- FIGURA 1: Boxplots con anotaciones ---
    fig, axes = plt.subplots(1, len(factor_cols), figsize=figsize, sharey=False)
    if len(factor_cols) == 1:
        axes = [axes]

    for ax, factor in zip(axes, factor_cols):
        group1 = df.loc[df[group_col] == group_labels[0], factor].dropna()
        group2 = df.loc[df[group_col] == group_labels[1], factor].dropna()

        # Mann-Whitney U test
        if len(group1) > 0 and len(group2) > 0:
            stat, pval = mannwhitneyu(group1, group2, alternative='two-sided')
        else:
            stat, pval = np.nan, np.nan

        med1 = group1.median() if len(group1)>0 else np.nan
        med2 = group2.median() if len(group2)>0 else np.nan
        diff_meds = med1 - med2

        resultados.append({
            "Factor": factor,
            "n_"+group_labels[0]: len(group1),
            "n_"+group_labels[1]: len(group2),
            "mean_"+group_labels[0]: group1.mean() if len(group1)>0 else np.nan,
            "mean_"+group_labels[1]: group2.mean() if len(group2)>0 else np.nan,
            "median_"+group_labels[0]: med1,
            "median_"+group_labels[1]: med2,
            "diff_medianas": diff_meds,
            "U_stat": stat,
            "p_valor": pval
        })

        # Boxplot
        sns.boxplot(x=group_col, y=factor, data=df, ax=ax, palette=["#5B9BD5","#F4B183"])
        ax.set_title(factor)
        ax.set_xlabel("")
        ax.set_ylabel("Score")

        # Anotar estadístico y p-value
        ymax = df[factor].max(skipna=True)
        yloc = ymax + 0.05*(abs(ymax) + 1e-6)
        text = f"U={stat:.0f}\np={pval:.2e}"
        ax.text(0.5, 1.05, text, ha='center', va='bottom', fontsize=9, transform=ax.transAxes)

    plt.tight_layout()
    if save_fig:
        plt.savefig(f"{output_prefix}_boxplots.png", dpi=300, bbox_inches="tight")
    plt.show()

    resumen_df = pd.DataFrame(resultados).set_index("Factor")

    # --- FIGURA 2: Barras con medias + IC bootstrap ---
    if show_bar_ci:
        n_boot = 1000
        fig2, axes2 = plt.subplots(1, len(factor_cols), figsize=figsize, sharey=False)
        if len(factor_cols) == 1:
            axes2 = [axes2]
        for ax, factor in zip(axes2, factor_cols):
            means = []
            cis_low = []
            cis_high = []
            for lab in group_labels:
                vals = df.loc[df[group_col] == lab, factor].dropna().values
                if len(vals) == 0:
                    means.append(np.nan); cis_low.append(np.nan); cis_high.append(np.nan); continue
                mean = vals.mean()
                boots = []
                for _ in range(n_boot):
                    sample = np.random.choice(vals, size=len(vals), replace=True)
                    boots.append(sample.mean())
                ci_low = np.percentile(boots, 2.5)
                ci_high = np.percentile(boots, 97.5)
                means.append(mean); cis_low.append(ci_low); cis_high.append(ci_high)
            x = np.arange(len(group_labels))
            ax.bar(x, means, yerr=[np.array(means)-np.array(cis_low), np.array(cis_high)-np.array(means)],
                   color=["#5B9BD5","#F4B183"], alpha=0.9, capsize=6)
            ax.set_xticks(x); ax.set_xticklabels(group_labels)
            ax.set_title(f"{factor} — Media ± IC 95%")
            ax.set_ylabel("Media del score")
        plt.tight_layout()
        if save_fig:
            plt.savefig(f"{output_prefix}_barplots.png", dpi=300, bbox_inches="tight")
        plt.show()

    return resumen_df

# Ejemplo de uso:
# resumen = summary_and_plots_with_stats(
#     df_scores,  # tu DataFrame con factores y columna 'Condicion' ('ON'/'OFF')
#     factor_cols=["Factor1","Factor3"],
#     group_col="Condicion",
#     group_labels=("ON","OFF"),
#     figsize=(12,5),
#     show_bar_ci=True,
#     save_fig=True,
#     output_prefix="factores_on_off"
# )
# print("\nTabla resumen (lista para diapositiva):")
# display(resumen.round(4))
```

```python
import matplotlib.pyplot as plt

def plot_factor_medians_by_station(resultados_finales, factores=["Factor1", "Factor3"]):
    """
    Grafica las medianas ON/OFF wind de los factores por estación usando el DataFrame resultados_finales.
    Muestra el valor p y U-stat encima de cada barra.
    """
    estaciones = resultados_finales["Estacion"].unique()
    for factor in factores:
        df_factor = resultados_finales[resultados_finales["Factor"] == factor]
        x = range(len(estaciones))
        med_on = df_factor["mediana_on"].values
        med_off = df_factor["mediana_off"].values
        pvals = df_factor["p_valor"].values
        u_stats = df_factor["U_stat"].values

        width = 0.35
        fig, ax = plt.subplots(figsize=(12, 5))
        bars_on = ax.bar([i - width/2 for i in x], med_on, width, label="ON WIND", color="skyblue")
        bars_off = ax.bar([i + width/2 for i in x], med_off, width, label="OFF WIND", color="salmon")

        # Anotar p-value y U-stat
        for i, (bar_on, bar_off, p, u) in enumerate(zip(bars_on, bars_off, pvals, u_stats)):
            y_max = max(bar_on.get_height(), bar_off.get_height())
            ax.text(i, y_max + 0.05*y_max, f"p={p:.2g}\nU={u:.0f}", ha="center", va="bottom", fontsize=9)

        ax.set_xticks(x)
        ax.set_xticklabels(estaciones, rotation=30, ha="right")
        ax.set_ylabel("Mediana del factor")
        ax.set_title(f"Mediana diaria ON vs OFF WIND — {factor}")
        ax.legend()
        plt.tight_layout()
        plt.show()

# Ejemplo de uso:
plot_factor_medians_by_station(resultados_finales, factores=["Factor1", "Factor3"])
```
